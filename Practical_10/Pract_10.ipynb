```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("default")
sns.set_theme()

df = pd.read_csv("pokemon.csv")

print("First 5 rows of dataset:")
display(df.head())

print("\nDataset Info:")
df.info()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(
    data=df,
    x="attack",
    y="defense",
    hue="is_legendary",
    style="is_legendary",
    alpha=0.6
)
plt.title("Attack vs Defense by Legendary Status")
plt.xlabel("Attack")
plt.ylabel("Defense")
plt.legend(title="Legendary (1=Yes, 0=No)")
plt.grid(True, linestyle="--", alpha=0.7)
plt.tight_layout()
plt.show()

print("Observation: Strong clusters may indicate differences between legendary and non-legendary Pokémon.")
```

```python
speed_gen = df.groupby("generation")["speed"].mean().reset_index()

plt.figure(figsize=(8,5))
sns.lineplot(data=speed_gen, x="generation", y="speed", marker="o", color="purple")
plt.title("Average Pokémon Speed by Generation")
plt.xlabel("Generation")
plt.ylabel("Average Speed")
plt.grid(True, linestyle="--", alpha=0.7)
plt.tight_layout()
plt.show()

print("Observation: Shows how Pokémon design changed across generations.")
```

```python
num_cols = ['hp','attack','defense','sp_attack','sp_defense','speed']

sns.pairplot(df[num_cols + ['is_legendary']], hue="is_legendary", palette="Set2", height=1.2)
plt.suptitle("Pairplot: Stats vs Legendary Status", y=1.02)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.violinplot(x="is_legendary", y="weight_kg", data=df, palette="Pastel1")
plt.title("Weight Distribution by Legendary Status")
plt.xlabel("Legendary (1=Yes)")
plt.ylabel("Weight (kg)")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df["speed"], bins=30, kde=True)
plt.title("Distribution of Pokémon Speed")
plt.xlabel("Speed")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df["attack"], bins=30, kde=True)
plt.title("Distribution of Pokémon Attack")
plt.xlabel("Attack")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.countplot(data=df, x="type1")
plt.title("Pokémon Count by Primary Type")
plt.xlabel("Type 1")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(6,5))
sns.countplot(data=df, x="generation")
plt.title("Pokémon Count by Generation")
plt.xlabel("Generation")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(5,5))
sns.countplot(data=df, x="is_legendary")
plt.title("Legendary vs Non-Legendary Pokémon")
plt.xlabel("Is Legendary")
plt.ylabel("Count")
plt.show()

legendary_rate = df["is_legendary"].mean()*100
print(f"Legendary Pokémon Rate: {legendary_rate:.2f}%")
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x="is_legendary", y="speed")
plt.title("Speed vs Legendary Status")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x="is_legendary", y="attack")
plt.title("Attack vs Legendary Status")
plt.show()
```

```python
type_rate = df.groupby("type1")["is_legendary"].mean().reset_index()
type_rate["is_legendary"] *= 100

display(type_rate)

plt.figure(figsize=(10,5))
sns.barplot(data=type_rate, x="type1", y="is_legendary")
plt.xticks(rotation=45)
plt.title("Legendary Rate by Pokémon Type")
plt.ylabel("Legendary Rate (%)")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.kdeplot(data=df, x="speed", hue="is_legendary", multiple="stack")
plt.title("Speed Distribution by Legendary Status")
plt.show()
```

```python
sns.jointplot(data=df, x="attack", y="defense", hue="is_legendary", kind="scatter", height=7)
plt.suptitle("Attack vs Defense by Legendary Status", y=1.02)
plt.show()
```

```python
type_counts = df.groupby(["type1","is_legendary"]).size().reset_index(name="Count")
type_pivot = type_counts.pivot(index="type1", columns="is_legendary", values="Count")
type_pivot.columns = ["Non-Legendary", "Legendary"]

display(type_pivot)

plt.figure(figsize=(10,5))
bottom = None
for col in type_pivot.columns:
    if bottom is None:
        plt.bar(type_pivot.index, type_pivot[col])
        bottom = type_pivot[col]
    else:
        plt.bar(type_pivot.index, type_pivot[col], bottom=bottom)
        bottom += type_pivot[col]

plt.xticks(rotation=45)
plt.title("Legendary Distribution Across Pokémon Types")
plt.ylabel("Count")
plt.legend(type_pivot.columns)
plt.show()
```

```python
num_df = df.select_dtypes(include=[np.number])

plt.figure(figsize=(12,7))
sns.heatmap(num_df.corr(), annot=False, cmap="coolwarm")
plt.title("Correlation Heatmap of Pokémon Stats")
plt.show()
```

```python
plt.figure(figsize=(12,6))

plt.subplot(1,3,1)
sns.histplot(df["attack"], bins=20, kde=True)
plt.title("Attack Distribution")

plt.subplot(1,3,2)
sns.countplot(data=df, x="generation")
plt.title("Count by Generation")

plt.subplot(1,3,3)
sns.countplot(data=df, x="is_legendary")
plt.title("Legendary Count")

plt.suptitle("Mixed Visualization", fontsize=16)
plt.tight_layout()
plt.show()
```

```python
legend_counts = df["is_legendary"].value_counts()

plt.figure(figsize=(5,5))
plt.pie(legend_counts, labels=["Non-Legendary", "Legendary"], autopct="%1.1f%%", pctdistance=0.85)
circle = plt.Circle((0,0), 0.60, fc="white")
plt.gca().add_artist(circle)
plt.title("Legendary Pokémon Donut Pie")
plt.show()
```

```python
plt.figure(figsize=(5,5))
plt.pie(legend_counts, labels=["Non-Legendary", "Legendary"], autopct="%1.1f%%")
plt.title("Distorted Pie Chart")
plt.gca().set_aspect("auto")
plt.show()
```

```python
plt.figure(figsize=(5,5))
plt.pie(legend_counts, labels=["Non-Legendary", "Legendary"], autopct="%1.1f%%", explode=[0.05,0.1])
plt.title("Exploded Pie Chart (Distorted)")
plt.show()
```
