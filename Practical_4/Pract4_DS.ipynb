**Aim:** Hypothesis Testing
* Formulate null and alternative hypotheses for a given problem.
* Conduct a hypothesis test using appropriate statistical tests (e.g., t-test chi-square test).
* Interpret the results and draw conclusions based on the test outcomes.

```
import pandas as pd
import scipy.stats as stats

# Load the dataset
df = pd.read_csv("pokemon.csv")

# Sample data: We will use the 'attack' column
data = df['attack']

print("First 10 Attack values : ", data.head(10).values)

# Define the null hypothesis
H0 = "The average Attack of Pokemon is 80."

# Define the alternative hypothesis
H1 = "The average Attack of Pokemon is not 80."

# Calculate the test statistic
# We perform a 1-sample t-test against the population mean of 80
t_stat, p_value = stats.ttest_1samp(data, 80)

# Print the results
print("Test statistic:", t_stat)
print("p-value:", p_value)

# Conclusion
if p_value < 0.05:
    print("Reject the null hypothesis.")
else:
    print("Fail to reject the null hypothesis.")
```

### **Two sampled T-test**

The Independent Samples t-test compares the means of two independent groups. Here, we will compare the **Attack** of **Legendary** Pokemon vs **Non-Legendary** Pokemon to see if there is a significant difference.

```
from scipy.stats import ttest_ind
import numpy as np

# Select two independent groups
# Compare Attack of Legendary (is_legendary=1) vs Non-Legendary (is_legendary=0)
legendary_attack = df[df['is_legendary'] == 1]['attack']
non_legendary_attack = df[df['is_legendary'] == 0]['attack']

print("Legendary Attack (first 5):", legendary_attack.head().values)
print("Non-Legendary Attack (first 5):", non_legendary_attack.head().values)

legendary_mean = np.mean(legendary_attack)
non_legendary_mean = np.mean(non_legendary_attack)

print("Legendary mean attack:", legendary_mean)
print("Non-Legendary mean attack:", non_legendary_mean)

legendary_std = np.std(legendary_attack)
non_legendary_std = np.std(non_legendary_attack)

print("Legendary std deviation:", legendary_std)
print("Non-Legendary std deviation:", non_legendary_std)

# Perform Independent t-test
ttest, pval = ttest_ind(legendary_attack, non_legendary_attack)
print("p-value", pval)

if pval < 0.05:
    print("We reject null hypothesis")
else:
    print("We accept null hypothesis")
```

### **Paired sampled t-test**

The paired sample t-test checks for differences between two related variables. We can compare **Attack** and **Defense** stats for the same Pokemon to see if they differ significantly on average.

**H0**: Mean difference between Attack and Defense is 0.
**H1**: Mean difference is not 0.

```
import pandas as pd
from scipy import stats
import numpy as np

# Select a subset for clearer demonstration
subset_df = df.head(100)

print(subset_df[['attack', 'defense']].describe())

# Perform Paired t-test
# We are comparing two features of the SAME entity (Pokemon), which makes it paired
ttest, pval = stats.ttest_rel(subset_df['attack'], subset_df['defense'])
print("p-value:", pval)

if pval < 0.05:
    print("Reject null hypothesis")
else:
    print("Accept null hypothesis")
```

### **When you can run a Z Test.**

We use a Z-test because our sample size is large (N \> 30). We will test if the mean **HP** is significantly different from 70.

**One-sample Z test**

```
import pandas as pd
from scipy import stats
from statsmodels.stats import weightstats as stests

# H0: The mean HP is 70
ztest, pval = stests.ztest(df['hp'], x2=None, value=70)

print(float(pval))

if pval < 0.05:
    print("Reject null hypothesis")
else:
    print("Accept null hypothesis")
```

### **Two-sample Z test**

Here we check two independent data groups. We will compare the **HP** of **Generation 1** Pokemon vs **Generation 2** Pokemon.

**H0**: Mean HP of Generation 1 and Generation 2 groups is equal.
**H1**: Mean HP is not equal.

```
# Selecting two independent groups
gen1_hp = df[df['generation'] == 1]['hp']
gen2_hp = df[df['generation'] == 2]['hp']

# Perform Two-sample Z-test
ztest, pval1 = stests.ztest(gen1_hp, x2=gen2_hp,
                            value=0, alternative='two-sided')

print(float(pval1))

if pval1 < 0.05:
    print("Reject null hypothesis")
else:
    print("Accept null hypothesis")
```

### **Chi-Square Test**

This test is applied to two categorical variables. We will determine if there is a significant association between **Generation** and **is\_legendary** status.

```
import pandas as pd
from scipy.stats import chi2_contingency

# Create a contingency table
contingency_table = pd.crosstab(df['generation'], df['is_legendary'])
print("Contingency Table:\n", contingency_table)

# Perform the chi-square test
chi2_statistic, p_value, dof, expected_frequencies = chi2_contingency(contingency_table)

# Print the results
print('Chi-square statistic:', chi2_statistic)
print('P-value:', p_value)
print('Degrees of freedom:', dof)
print('Expected frequencies:')
print(expected_frequencies)

if p_value < 0.05:
    print("Reject null hypothesis (Variables are related)")
else:
    print("Accept null hypothesis (Variables are independent)")
```
