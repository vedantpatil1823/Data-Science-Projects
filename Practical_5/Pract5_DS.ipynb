### **ANOVA (F-TEST)**

**One Way F-test (Anova):** We will compare the means of **Attack** across different **Primary Types** (Fire, Water, Grass) to see if they are similar.

```python
import pandas as pd
import scipy.stats as stats

# Load the dataset
df = pd.read_csv("pokemon.csv")

# Sample data: We will use the 'attack' column
data = df['attack']

# Prepare data for ANOVA
# We select a subset of types for clarity
types_to_compare = ['fire', 'water', 'grass']
d_data = {grp: df['attack'][df.type1 == grp] for grp in types_to_compare}

# Perform One-way ANOVA
F, p = stats.f_oneway(d_data['fire'], d_data['water'], d_data['grass'])

print("p-value for significance is: ", p)

if p < 0.05:
    print("Reject null hypothesis")
else:
    print("Accept null hypothesis")
```

### **Two Way F-test**

We will check the effect of two independent variables (**is\_legendary** and **generation**) on a dependent variable (**base\_total**).

```python
import statsmodels.api as sm
from statsmodels.formula.api import ols

# Fit the OLS model
# We are testing: base_total ~ is_legendary + generation + is_legendary*generation
# Note: 'generation' is categorical here, so we wrap it in C()
model = ols('base_total ~ C(is_legendary)*C(generation)', df).fit()

print(f"Overall model F({model.df_model: .0f},{model.df_resid: .0f}) = {model.fvalue: .3f}, p = {model.f_pvalue: .4f}")

# Perform ANOVA on the model
res = sm.stats.anova_lm(model, typ=2)
print(res)
```
